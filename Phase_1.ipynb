basic of numpy

```python
import numpy as np
```

list vs array

```python
py_list = [1,2,3]
print("pyhon list multiplication",py_list * 2)
```

```python
np_array =np.array ([1,2,3]) #element wise multiplication
print("pyhon array multiplication",np_array * 2)

import time
start = time.time()
py_list = [i*2 for i in range(10000000)]
print("\n list operation time ",time.time()-start)

start = time.time()
np_array = np.arange(10000000)
print("\n numpy array operation time ",time.time()-start)
```

Creating Array From Scratch

```python
zeros = np.zeros((3,4))
print("zeros array:\n",zeros)

ones = np.ones((3,4))
print("ones array:\n",ones)

full = np.full((3,4),5)
print("full array:\n",full)

random = np.random.random((3,4))
print("random array:\n",random * 100)
sequence = np.arange(0,10,2)
print("sequence array:\n",sequence)
```

vector matrix and tensor

```python
vector = np.array([1,2,3,4])
print("vector:",vector)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print("matrix :\n",matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("tensor :\n",tensor)
```

Array Properties

```python
arr = np.array([[1,2,3],[4,5,6]])
print("array is:\n",arr)
print("shape:",arr.shape)
print("dimension:",arr.ndim)
print("size:",arr.size)
print("data type:",arr.dtype)
```

Array reshaping

```python
arr = np.arange(12)
print("original array:\n",arr)

reshaped = arr.reshape((3,4))
print("reshaped array:\n",reshaped)

flattened = reshaped.flatten()
print("flattened array:\n",flattened)

# return view, intead of copy(original array)
raveled = reshaped.ravel()
print("raveled array:\n",raveled)

#transpose
transpose = reshaped.T
print("transpose array:\n",transpose)

```
